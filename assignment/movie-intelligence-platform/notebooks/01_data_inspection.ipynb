```python
# Import the libraries
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully!")
```

```python
# Load the dataset
df = pd.read_csv("../data/raw/movies.csv")

print("Dataset loaded successfully!")
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.head()
```

```python
df.columns.tolist()
```

```python
df.info()
```

```python
# Missing values
missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

missing
```

```python
missing_percentage = (
    df.isnull().mean() * 100
).sort_values(ascending=False)

missing_percentage = missing_percentage[missing_percentage > 0]

missing_percentage.round(2)
```

```python
# Duplicate movies
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate movie names:", df["name"].duplicated().sum()) 
```

```python
# Numerical statistics
df.describe()
```

```python
# Categorical columns 
categorical_columns = [
    "rating",
    "genre",
    "director",
    "writer",
    "star",
    "country",
    "company"
]

for column in categorical_columns:
    print(f"\n===== {column.upper()} =====")
    print("Unique values:", df[column].nunique())
    print(df[column].value_counts(dropna=False).head(10))
```

```python
# Numerical columns 
numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

print("Numerical columns:")
print(numerical_columns.tolist()) 
```

```python
# Check suspicious values
print("Budget statistics:")
print(df["budget"].describe())

print("\nGross statistics:")
print(df["gross"].describe()) 
```

```python
# Inspect runtime 
print(df["runtime,,"].head(20))
print(df["runtime,,"].dtype)
```

```python
# Check years 
print("Minimum year:", df["year"].min())
print("Maximum year:", df["year"].max())

print("\nYear value counts:")
print(df["year"].value_counts().sort_index().head(20))
```

```python
# Quick data-quality report
quality_report = pd.DataFrame({
    "column": df.columns,
    "dtype": df.dtypes.astype(str).values,
    "missing_count": df.isnull().sum().values,
    "missing_percentage": (
        df.isnull().mean().values * 100
    ).round(2),
    "unique_values": df.nunique().values
})

quality_report
```
